# 07 Plotly Dash — reproducible offline validation

This notebook checks the supplied course data. It does not claim to rerun external API/scraping workflows.

```python
from pathlib import Path
import pandas as pd
import plotly.express as px
app_path=Path('spacex_dash_app.py')
print('Dash application source exists:',app_path.exists())
df=pd.read_csv('data/dataset_part_2.csv')
site_success=df.groupby('LaunchSite',as_index=False)['Class'].sum()
fig=px.pie(site_success,values='Class',names='LaunchSite',title='Successful landings by launch site')
fig.write_html('outputs/dash_all_sites.html')
display(fig)
print('Static Plotly figure generated. Full Dash server interactions not tested in notebook.')
```
