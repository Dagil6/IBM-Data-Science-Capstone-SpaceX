# Folium — launch sites, outcomes and distances
Interactive maps and site proximity analysis.

```python
import pandas as pd, folium
from folium.plugins import MarkerCluster
from pathlib import Path
D=Path('../data');O=Path('../outputs');O.mkdir(exist_ok=True)
df=pd.read_csv(D/'dataset_part_2.csv')
coords=df.dropna(subset=['Latitude','Longitude']).copy()
if coords.empty: raise ValueError('Coordinates absent: merge site coordinates from the API notebook first')
m=folium.Map(location=[29,-95],zoom_start=3)
for name,g in coords.groupby('LaunchSite'):
    lat,lon=g[['Latitude','Longitude']].iloc[0]
    folium.Marker([lat,lon],tooltip=f'{name} ({len(g)} flights)').add_to(m)
m.save(O/'folium_sites.html');display(m)
```

```python
m2=folium.Map(location=[29,-95],zoom_start=3)
cluster=MarkerCluster().add_to(m2)
for _,r in coords.iterrows():
    color='green' if r['Class']==1 else 'red'
    folium.CircleMarker([r['Latitude'],r['Longitude']],radius=4,color=color,fill=True,
        popup=f"{r['LaunchSite']}: {'success' if r['Class']==1 else 'failure'}").add_to(cluster)
m2.save(O/'folium_outcomes.html');display(m2)
```

**Proximity analysis:** For the rubric, open a selected launch site in Folium and use measured coastline, highway and railway coordinates (for example from OpenStreetMap). Do not invent distances. The following cell computes great-circle distances once coordinates are verified.

```python
from math import radians,sin,cos,asin,sqrt
def distance_km(a,b):
    lat1,lon1=map(radians,a);lat2,lon2=map(radians,b)
    x=sin((lat2-lat1)/2)**2+cos(lat1)*cos(lat2)*sin((lon2-lon1)/2)**2
    return 6371*2*asin(sqrt(x))
print('Provide verified [lat,lon] coordinates for coast, rail and highway, then use distance_km(site, point).')
```
